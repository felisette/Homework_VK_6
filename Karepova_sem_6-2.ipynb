### Задание 1. 
Полное бинарное дерево

Реализуйте функцию, которая проверяет, является ли заданное бинарное дерево полным. Необходимо корректно определить условие заполненности уровней и объяснить выбранный способ обхода (например, BFS). 
Ответ: Выбор обход в ширину - проверяем, что все уровни заплнены полностью, а последний слева направо без пропусков, что обеспечивает полноту дерева. Обход в глубину не удобен, так как не рассматривает уровни отдельно и потребовалось бы хранить глубину.

```python
class TreeNode:
    def __init__(self, val = 0, left = None, right = None):
        self.data = val
        self.left = left
        self.right = right
```

```python
def buildTree(arr, i):
    if i >= len(arr) or arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = buildTree(arr, 2*i + 1)
    root.right = buildTree(arr, 2*i + 2)
    return root

def printTree(node):
    if node:
        printTree(node.left)
        print(node.data, end=" ")
        printTree(node.right)

def print_tree(node, prefix="", is_left=True, is_root=True):
    if node is None:
        return
    if node.right is not None:
        print_tree(node.right, prefix + ("    " if is_root or not is_left else "│   "), False, False)
    if is_root:
        print(f"{node.data}")
    else:
        print(f"{prefix}{'└── ' if is_left else '┌── '}{node.data}")
    if node.left is not None:
        print_tree(node.left, prefix + ("    " if is_root or is_left else "│   "), True, False)
```

```python
def isCompleteTree(root):
    if root==None:
        return True
    queue = deque([root])
    seenNull = False
    while queue:
        node = queue.popleft()
        if node == None:
            seenNull = True
        else:
            if seenNull:
                return False
            queue.append(node.left)
            queue.append(node.right)
    return True
```

```python
val  = [21, 19, 18, 11, 12, 15, 9, 8, 10]
root = buildTree(val, 0)
print_tree(root)
print(isCompleteTree(root))
arr = [1, 2, 3, None, None, None, 7]
root = buildTree(arr, 0)
print_tree(root)
print(isCompleteTree(root))
```

### Задание 2 
Объединение K отсортированных массивов

Проанализируйте сложность алгоритма: сложность O(K + NlogK), потому что O(K) - инициализация кучи, O(NlogK) - извлечение из кучи. K - число массивов, N - общее число элементов в массивах

```python
def mergeKSortedArrays(sortedArrays):
    merged_array = []
    min_heap = []
    for i in range(len(sortedArrays)):
        current_array = sortedArrays[i]
        if len(current_array) > 0:
            heapq.heappush(min_heap, (current_array[0], i, 0))
    while min_heap:
        value, array_idx, element_idx = heapq.heappop(min_heap)
        merged_array.append(value)
        if element_idx + 1 < len(sortedArrays[array_idx]):
            next_element = sortedArrays[array_idx][element_idx + 1]
            heapq.heappush(min_heap, (next_element, array_idx, element_idx + 1))
    return merged_array
```

```python
arr1 = [[1, 4, 5], [1, 3, 4], [2, 6]]
print(mergeKSortedArrays(arr1)) 
arr2 = [[-10, -5, 0], [-7, -2, 3], [-8, 1]]
print(mergeKSortedArrays(arr2)) 
arr3 = [[1, 5, 9], [], [2, 3, 4]]
print(mergeKSortedArrays(arr3)) 
arr4 = [[], [], []]
print(mergeKSortedArrays(arr4)) 
```

### Задание 3
K-ый наименьший / наибольший элемент

Сложность алгоритма: O(N*logK), где N - число элементов в массиве, K - номер искомого элемента, потому что проходим массив за O(N), вставка и удаление из кучи происходит за O(logK)

```python
def kMin(nums, k):
    max_heap = []
    for num in nums:
        heapq.heappush(max_heap, -num) # берем с минусом, потому что по умолчанию строится MinHeap
        if len(max_heap) > k:
            heapq.heappop(max_heap)
    return -max_heap[0]

def kMax(nums, k):
    min_heap = []
    for num in nums:
        heapq.heappush(min_heap, num)
        if len(min_heap) > k:
            heapq.heappop(min_heap)
    return min_heap[0]
```

```python
val = [9, 8, 8, 3, 6, 6, 3]
print(kMin(val, 3), kMax(val, 3))

val = [3, 7, 8, 9, 6, 9]
print(kMin(val, 1), kMax(val, 1))

val = [11, 6, 16, 2, 9]
print(kMin(val, 4), kMax(val, 4))
```

### Задание 4
K-ый наименьший / наибольший элемент в BST

```python
def inorderMin(node, k, counter):
    if node is None:
        return None
    left_result = inorderMin(node.left, k, counter)
    if left_result is not None:
        return left_result
    counter[0] += 1
    if counter[0] == k:
        return node.data
    return inorderMin(node.right, k, counter)

def inorderMax(node, k, counter):
    if node is None:
        return None
    right_result = inorderMax(node.right, k, counter)
    if right_result is not None:
        return right_result
    counter[0] += 1
    if counter[0] == k:
        return node.data
    return inorderMax(node.left, k, counter)
```

```python
data = [4, 2, 6, 1, 3, 5, 7]
root = buildTree(data, 0)
print_tree(root)
print(inorderMin(root, 3, [0]))
print(inorderMax(root, 3, [0]))
```

### Задание 5
Balance factor

Высота узла вычисляется как сумма 1 наибольшей из высот левого и праввого поддерева, при этом высота пустого узла равна 0

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right
        self.balanceFactor = 0

def buildTree(arr, i):
    if i >= len(arr) or arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = buildTree(arr, 2*i + 1)
    root.right = buildTree(arr, 2*i + 2)
    return root
```

```python
def calculateHeightsAndBalance(node):
    if not node:
        return 0
    leftHeight = calculateHeightsAndBalance(node.left)
    rightHeight = calculateHeightsAndBalance(node.right)
    node.balanceFactor = leftHeight - rightHeight
    return 1 + max(leftHeight, rightHeight)
```

```python
arr = [4, 2, 6, 1, 3, 5, 7]
root = buildTree(arr, 0)
print(calculateHeightsAndBalance(root))
```

### Задание 6
Преобразование в зеркальное дерево

Алгоритм корректен, потому что при обходе меняет местами правый и левый потомок для каждого узла, потом рекурсивно вызывается от потомков, что обеспечивает полное зеркально отражение дерева. 

Временная сложность O(N), где N - число узлов в дереве, так как мы посещаем каждый узел один раз и делаем операцию с ним за O(1).

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right

def buildTree(arr, i):
    if i >= len(arr) or arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = buildTree(arr, 2*i + 1)
    root.right = buildTree(arr, 2*i + 2)
    return root
```

```python
def mirrorTree(node):
    if node is None:
        return None
    node.left, node.right = node.right, node.left
    mirrorTree(node.left)
    mirrorTree(node.right)
    return node
```

```python
arr = [4, 2, 6, 1, 3, 5, 7]
root = buildTree(arr, 0)
print_tree(root)

mirrorTree(root)
print_tree(root)
```
